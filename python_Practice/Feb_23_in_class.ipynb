```python
# while loop
r = 1
n = 9

while n > 0:    
    r = r * n    
    n = n - 1    
    print(r, n, n > 0)

#loop takes the recent results, n and r, and re-inputted into the conditions
## first while loop: r = 9, n = 8
```

```python
# while loop
n = 9
r = 1
while n > 0:    
    print(r, n, n > 0)
    r = r * n    # r = 9
    n = n - 1    # n = 8

#after the first while loop, the print shows up before the r and n are carried, thus shows up in the second while loop
## the last loop,the statement n>0 is returned True because it only shows the n and r from its prior execution
```

```python
# if-elif-else statement
import random
n = random.randint(1,100)
print(n)
if n > 80:
    print("Greater than 80")
elif n < 20:
    print("Less than 20")
else:
    print("Between 20 and 80")
```

```python
# Write a for loop to iterate through the list my_list and print the 
# current item's value and "is positive" or "is negative," depending on its value

my_list = [-17, 5, 90, -3.5, 12, -23, 0]

for i in my_list:
    if i > 0:
        print(f"{i} is positive.")
    elif i < 0:
        print(f"{i} is negative.")
    else:
        print(f"{i} is zero.")
        
```
