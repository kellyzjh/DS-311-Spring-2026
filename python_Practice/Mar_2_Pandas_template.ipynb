### Adding Rows

```python
# we forgot 2 extra rows from our DF. How do we add them?
extra_rows = {'age': [40, 30], 
              'sex': [1, 0], 
              'cp': [0, 0], 
              'trestbps': [120, 130],
              'chol': [240, 200],
              'fbs': [0, 0], 
              'restecg': [1, 0], 
              'thalach': [120, 122], 
              'exang': [0, 1],
              'oldpeak': [0.1, 1.0], 
              'slope': [1, 1], 
              'ca': [0, 1], 
              'thal': [2, 3],
              'target': [0, 0]}
extra_rows

# convert the extra_rows to a DF
missing = pd.DataFrame(extra_rows)
missing
```

```python
# get help on the pd.concat() method to understand the syntax
heart_augmented = pd.concat([heart_df, missing],
                           ignore_index=True)
```

### Adding Columns

```python
# add a column called "test" to the DF and fill it with zeros
heart_df["test column"] = 0
heart_df["test column"]
```

```python
# add a column called "age+cholesterol", where you add 
# the age and cholesterol values together

heart_df["age+cholesterol"] = heart_df["age"] + heart_df["chol"]

heart_df["age+cholesterol"]
```

### Filtering

```python
heart_augmented['age'] >= 70
```

```python
heart_augmented[(heart_augmented['age'] >= 70) |
                (heart_augmented['sex'] == 1) &
                (heart_augmented['fbs'] == 0)]
```

```python
# Display the patients who are 70 or over 
# as well as the patients whose trestbps score is greater than 170.

heart_augmented[(heart_augmented['age'] >= 70) | (heart_augmented['trestbps'] >= 170)]

```
