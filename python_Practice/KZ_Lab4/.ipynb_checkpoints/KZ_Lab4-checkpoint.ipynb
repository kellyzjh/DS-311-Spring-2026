## 3a) Number of fast food restaurants in each provice and city

```python
num_ff_by_prov_city = ff_df.groupby(["province","city"])[["id"]].count().rename(columns={"id": "num_fastfood"})

num_ff_by_prov_city
```

## 3b) Top 5 provinces with most fastfood restaurants

```python
top_5prov = ff_df.groupby(["province"])[["id"]].count().rename(columns={"id": "num_fastfood"}).sort_values(by="num_fastfood", ascending=False).head(5)

top_5prov

## The top 5 provinces with the most fast food restaurants are
## California, Texas, Florida, Ohio, and Georgia, descending
```

## 3c) fastfood restaurants in Hawaii

```python
hawaii_ff = ff_df[ff_df["province"] == "HI"].groupby("name").count().rename(columns={"id": "num_fastfood"})
hawaii_ff = hawaii_ff[["num_fastfood"]] 

hawaii_ff

## The fast food restaruants in Hawaii are 
## Big Island Pizza, Burger King, Chun Wah Kam Noodle Factory, Cupie's Drive-In, 
## Dairy Queen, Jack in the Box, McDonald's, Megs Drive In, Taco Bell, and Wendy's
```

## 3d) Most popular fastfood restaurant

```python
most_pop_ff = ff_df.groupby("name")[["id"]].count().rename(columns={"id": "num_fastfood"}).sort_values(by="num_fastfood", ascending=False)
most_pop_ff

# McDonald's is the most popular fastfood restaurant as it has the most restaurants 
```

## 3e) Year with most fast food restaurants added in CA

```python
ff_df["year"] = ff_df.dateAdded.str[:4]

CA_ff = ff_df[ff_df["province"] == "CA"]

CA_most_ff_added_year = CA_ff.groupby("year")[["id"]].count().rename(columns={"id": "num_fastfood"}).sort_values(by="num_fastfood", ascending=False)

CA_most_ff_added_year

## The year with most fast food restaurants added in CA is 2016. 
```
