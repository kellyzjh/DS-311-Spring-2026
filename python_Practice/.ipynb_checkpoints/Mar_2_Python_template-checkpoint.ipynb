## Lists

```python
# define list x
x = [2, "two", [1, 2, 3], 22]

# access the second element of x
x[1]
```

```python
# set y to contain the 3rd and 4th elements of x

y = x[2:4]
y
```

```python
# access the 2nd element using a negative index
x[-3]
```

```python
# create two new lists, each containing half of x

size = len(x)

first_half = x[0:size//2]
second_half = x[size//2:size]

first_half
second_half
```

```python
# generate a list y that is 4 repeats of [0,1]
y = 4*[0,1]
y
```

```python
# Modifying list elements
x = [0, 1, 2, 3, 4]

# replace '1' and '2' with the corresponding words
x[1:3] = ['one', 'two']

x
```

```python
# Add the number 7 as the new second item in the list
x[1:1]= [7]
x
```

```python
# remove the 7 from the list again
x[1:2] = []
x
```

```python
# append() vs. extend(): try append()
x = [1, 2, 3, 4]
y = [5, 6, 7]

x.append(y)

x
```

```python
# append() vs. extend(): try extend()
x = [1, 2, 3, 4]
y = [5, 6, 7]


x.extend(y)
x
```

```python
# using an if statement to remove an item
x = [10, 1.3, -2, 3, 14, -12]
if -2 in x:
    x.remove(-2)
x
```

### Sorting Lists

```python
# sorting lists
x = [10, 1.3, -2, 3, 14]

# created a list y that is the sorted version of x
y = sorted(x)

# print out both for comparison
y
```

```python
# Suppose there is a list and each element in the list is also a list,
# [[1,2,3], [2,1,3], [4,0,1]]
# What does the reverse-sorted list look like?

x = [[1,2,3], [2,1,3], [4,0,1]] #reverses the list

x.sort(reverse = True)
x
```

```python
# If we need to sort the order by the second element in each list,
# how should we use the key function to do it?

x = [[1,2,3], [2,1,3], [4,0,1]]
x.sort(key=lambda x:x[1])

"""
A lambda function is a small anonymous function that evaluate a 
single expression:
y = lambda arguments : expression

For example:
y = lambda a : a + 10
print(y(5))

The lambda function y takes the input (argument) 5, 
and returns 5 + 10

In our example above,
key = lambda x : x[1]

We are sorting based on whatever is in x[1].
"""
```

```python
# Assigning lists
x = [10, 1.3, -2, 3, 14]
y = x   #this saves y the same as x

# update y by changing the 3rd element to be -4
# print out both x and y

y = x[:]

y[2] = -4

print(x)
print(y)

```

```python
# how to fix this problem?
#to prevent x and y being the same, set y = x[:] instead of indicating they are the same variable
```

## Tuples

```python
x = (1, 2, 3)
x = x + (4,)
x
```

```python
# How to sort a tuple
x = (3, 1, 4, 2)

sort_x = sorted(x) #use sorted function returns as a new sorted list
sort_tuple = tuple(sort_x)

#OR 

sort_tuple = tuple(sorted(x))

sort_tuple
```

```python
x = (1, 2, 3, 4)
a, b, *c = x  #the asterisk (*) is used to collect multiple values into a lis
a, b, c
```

```python
*a, b, c = x
a, b, c 
```

```python
a, b, c, d, *e = x
a, b, c, d, e  # return (1, 2, 3, 4, [])
```

## Sets

```python
x = {1, 2, 2, 1, 2, 1, 2} 
x

x = {[1, 2, 2, 1, 2, 1, 2]}  #lists are mutuable hence cannot be elements of a set
x

my_list = [1, 2, 2, 1, 2, 1, 2]
x = set(my_list)
x
```

```python
x = {1, 2, 6}
y = {1, 7, 9}

check = 3 in x 
check
# try 3 in x
```

```python
# try 3 in x | y

check = 3 in x | y
check 
```

```python
# try 2 in x | y

check = 2 in x | y
check
```

```python
# try x & y

result = x & y #union
result
```

```python
# try x ^ y

result = x ^ y  #not union
result
```

```python
print(english_to_french.get("yellow"))
```

```python
print(english_to_french.get("Yellow", "Not a color"))
```
