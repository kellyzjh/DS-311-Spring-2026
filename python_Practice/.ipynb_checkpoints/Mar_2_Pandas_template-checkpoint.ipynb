### Adding Rows

```python
# we forgot 2 extra rows from our DF. How do we add them?
extra_rows = {'age': [40, 30], 
              'sex': [1, 0], 
              'cp': [0, 0], 
              'trestbps': [120, 130],
              'chol': [240, 200],
              'fbs': [0, 0], 
              'restecg': [1, 0], 
              'thalach': [120, 122], 
              'exang': [0, 1],
              'oldpeak': [0.1, 1.0], 
              'slope': [1, 1], 
              'ca': [0, 1], 
              'thal': [2, 3],
              'target': [0, 0]}
extra_rows

# convert the extra_rows to a DF
missing = pd.DataFrame(extra_rows)
missing
```

```python
# get help on the pd.concat() method to understand the syntax
heart_augmented = pd.concat([heart_df, missing],
                           ignore_index=True)
```
