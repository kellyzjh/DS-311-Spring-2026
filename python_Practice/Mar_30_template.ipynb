### Engineering some data

```python
# How many different genders are represented in our dataset?
```

```python
# Function to pare down the number of sexes
def sex_mapper(status):
    if status in ['Neutered Male', 'Intact Male']:
        return 'Male'
    elif status in ['Spayed Female', 'Intact Female']:
        return 'Female'
    else:
        return 'Unknown'
```

```python
animals_df['new_sex1'] = animals_df['sex_upon_outcome'].map(sex_mapper)
animals_df[:, ['sex_upon_outcome', 'new_sex1']]
```

### Filling in missing values
#### With np.nan or "Unknown"

```python
# Let's remember which columns said there were a lot of missing values
```

```python
# Create a copy of the data
animals_clean = animals_df.copy()

# insert np.nan for missing values
animals_name_filled = animals_clean.fillna('np.nan')
animals_name_filled.head()
```

```python
# Alternative way to fill missing values by specifying column(s) first
animals_only_names = animals_clean[['name']].fillna(value='UNKNOWN')
animals_only_names.head(5)
```

Question: Why are we using animals_df[['name']] in the first line of the code rather than just animals_df['name']?

```python
# Overwrite the existing 'name' column with the new one
animals_clean[['name']] = animals_only_names
animals_clean.head()
```

#### Or with imputed values

```python
outcome_subtype_counts = animals_clean['outcome_subtype'].value_counts()
outcome_subtype_counts
```

```python
# Gets only the categories in order of most frequent to least frequent
outcome_subtype_ordered = outcome_subtype_counts.index
print(outcome_subtype_ordered)

# And now find the most common one
most_common_outcome_subtype = outcome_subtype_ordered[0]
print(most_common_outcome_subtype)
```

```python
# Another way to find this is to use the .mode() function
print(animals_df['outcome_subtype'].mode())

# Note this is Series so we have to get the first element (which is the value)
most_common_outcome_subtype = animals_clean['outcome_subtype'].mode()[0]
most_common_outcome_subtype
```

```python
# Similar to the previous subsection, we can use fillna() and update the DF
animals_clean['outcome_subtype'] = animals_clean['outcome_subtype']\
    .fillna(most_common_outcome_subtype)
animals_clean.head()
```
