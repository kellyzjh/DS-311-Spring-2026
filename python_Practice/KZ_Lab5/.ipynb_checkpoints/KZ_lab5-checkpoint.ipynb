```python
import pandas as pd
import sqlite3 as sql
from matplotlib import pyplot as plt
import seaborn as sns
import numpy as np
```

```python
conn = sql.connect('./data/python.db')
ff_df = pd.read_sql_query("SELECT * FROM fastfood ", conn)
ff_df
```

### 4) This code takes the "postalCode" column containing string and creates a new column called "postalCodeNew" that turns the first five characters into integers by using the map function to iterate through the list with lambda x defined with the rules for the items going through the list. 

```python
ff_df["postalCodeNew"] = ff_df["postalCode"].map(lambda x:int(x[:5]))

ff_df
```

### 5) Restaurant chain with the greatest representation in SF, CA

```python

filter_SF = ff_df["postalCodeNew"].between(94100, 94188)
SF_rest_query = ff_df[filter_SF]

SF_rest_query = (SF_rest_query.groupby(["name", "postalCodeNew"])
                 .size()
                 .reset_index(name='num')
                 .sort_values(by='name', ascending=False))

SF_rest_query

## Subway has the greatest representation in SF,CA as this chain appears the most times and in different zipcodes throughout the city.
```

### 6g) Make up two sets of data and determine code to return 

```python
dataset =  {'price range':['5-10', '10-20', '20-30', '30-50', '50-100'], 
             'rating':[2.0, 3.5, 4.0, 4.1, 4.3]} 

dataset_df = pd.DataFrame(dataset, index=['id_1', 'id_2', 'id_3', 'id_4', 'id_5'])

dataset_df
```

```python
dataset_df.loc[dataset_df["rating"] >= 4.0]
#returns the restaurant price ranges with ratings equal to over 4.0
```

### 8) Generate bar chart after filtering

```python
#generate a list of restaurants that have at least 50 locations
counts = ff_df['name'].value_counts()
high_counts = counts[counts >= 50]
```

```python
#create a bar chart of these restaurants
_, ax = plt.subplots(figsize=(15, 6))
ax.bar(x=high_counts.index, height=high_counts)
ax.set_xticklabels(labels=high_counts.index, rotation=80);
```

### 9) Scatterplot

```python
buffalo = ff_df[ff_df['city'] == 'Buffalo']
_, ax = plt.subplots(figsize=(8, 8))
sns.scatterplot(data=buffalo,
x=buffalo['longitude'],
y=buffalo['latitude'],
hue=buffalo['name'],
ax=ax) ; 
```
