### Loops and conditional statements

```python
# for loop
b = [10, 5, 2]
for i in b:    
	print(i)

# Let's make it fancier by outputting 1/b[i]
```

```python
# while loop
n = 9
r = 1
while n > 0:    
    r = r * n    
    n = n - 1    
    print(r, n)
```

```python
# if-elif-else statement
import random
n = random.randint(1,100)
print(n)
if n > 80:
    print("Greater than 80")
elif n < 20:
    print("Less than 20")
else:
    print("Between 20 and 80")
```
